## 1. Data

```python
# 2026년 8월 수출입 동향
# https://www.motir.go.kr/kor/article/ATCL3f49a5a8c/172145/view

Month = [4, 5, 6, 7, 8]
DRAM_ASP= [35.0,37.5,40.0,45.0,46.5]
NAND_ASP= [24.2, 26.5, 28.8, 30.1, 30.5]
```


## 2. Linear Regression Analysis for DRAM_ASP

We will perform a linear regression analysis using `Month` as the independent variable and `DRAM_ASP` as the dependent variable.
After training the model, we will predict the `DRAM_ASP` for `Month = 9`.


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
import numpy as np

# Define independent and dependent variables
x_month = np.array(Month).reshape(-1, 1) # Independent variable (Month)
y_dram_asp = np.array(DRAM_ASP) # Dependent variable (DRAM_ASP)

# Create and train the linear regression model
model_dram = LinearRegression()
model_dram.fit(x_month, y_dram_asp)

# Calculate predicted values for the existing data
y_dram_pred = model_dram.predict(x_month)


# Predict DRAM_ASP for Month = 9
x_new_month = np.array([[9]])
y_dram_9_pred = model_dram.predict(x_new_month)
DRAM_ASP.append(y_dram_9_pred[0])
print(f"Predicted DRAM_ASP for Month 9: {y_dram_9_pred[0]:.2f}")

# Visualize the regression line
fig = plt.figure(figsize=(8, 6))
plt.scatter(x_month, y_dram_asp, color='blue', label='Original DRAM_ASP Data')
plt.plot(x_month, y_dram_pred, color='red', label='Regression Line')
plt.scatter(x_new_month, y_dram_9_pred, color='green', marker='X', s=100, label=f'Predicted (Month 9: {y_dram_9_pred[0]:.2f})')
plt.title('Linear Regression Analysis of DRAM_ASP vs. Month')
plt.xlabel('Month')
plt.ylabel('DRAM_ASP')
plt.legend()
plt.grid(True)
plt.show()
```


## 3. Linear Regression Analysis for NAND_ASP

Now, let's perform a linear regression analysis using `Month` as the independent variable and `NAND_ASP` as the dependent variable.
We will train a new model and predict the `NAND_ASP` for `Month = 9`.


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
import numpy as np

# Define independent and dependent variables
x_month_nand = np.array(Month).reshape(-1, 1) # Independent variable (Month)
y_nand_asp = np.array(NAND_ASP) # Dependent variable (NAND_ASP)

# Create and train the linear regression model for NAND_ASP
model_nand = LinearRegression()
model_nand.fit(x_month_nand, y_nand_asp)

# Calculate predicted values for the existing data
y_nand_pred = model_nand.predict(x_month_nand)


# Predict NAND_ASP for Month = 9
x_new_month_nand = np.array([[9]])
y_nand_9_pred = model_nand.predict(x_new_month_nand)
NAND_ASP.append(y_nand_9_pred[0])
print(f"Predicted NAND_ASP for Month 9: {y_nand_9_pred[0]:.2f}")

# Visualize the regression line
fig = plt.figure(figsize=(8, 6))
plt.scatter(x_month_nand, y_nand_asp, color='blue', label='Original NAND_ASP Data')
plt.plot(x_month_nand, y_nand_pred, color='red', label='Regression Line')
plt.scatter(x_new_month_nand, y_nand_9_pred, color='green', marker='X', s=100, label=f'Predicted (Month 9: {y_nand_9_pred[0]:.2f})')
plt.title('Linear Regression Analysis of NAND_ASP vs. Month')
plt.xlabel('Month')
plt.ylabel('NAND_ASP')
plt.legend()
plt.grid(True)
plt.show()
```

## 4. Calculating Growth Rate of DRAM ASP and NAND, using predicted data on September.

```python
q2_dram_asp = np.mean(DRAM_ASP[:3])
q3_dram_asp = np.mean(DRAM_ASP[3:])
q2_nand_asp = np.mean(NAND_ASP[:3])
q3_nand_asp = np.mean(NAND_ASP[3:])

grwoth_rate_dram_asp = (q3_dram_asp - q2_dram_asp) / q2_dram_asp *100
grwoth_rate_nand_asp = (q3_nand_asp - q2_nand_asp) / q2_nand_asp *100
print(f"Growth Rate of DRAM_ASP: {grwoth_rate_dram_asp:.2f}%")
print(f"Growth Rate of NAND_ASP: {grwoth_rate_nand_asp:.2f}%")
```
